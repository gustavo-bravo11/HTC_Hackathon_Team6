```python
data = pd.read_csv("../data_sets/Geoscout Facility Data Dump.csv", low_memory=False)
```

```python
# Skeleton for building the tables

# First column contains the primary key

data.columns = data.columns.str.replace('Unique Facility ID', 'id')

# Facility data
facilities = data.iloc[:, :13].drop_duplicates().reset_index(drop=True)

# Capacity data
facility_capacity = data.iloc[:, [0] + list(range(13, 35))].drop_duplicates().reset_index(drop=True)

# Production data

# I believe this table includes total production, but might need to take a look at it with the next table
facility_total_production = data.iloc[:, [0] + list(range(35, 45))].drop_duplicates().reset_index(drop=True)

# This is MONTHLY production data for that facility by LSD, month, and year, also explains the facility type
facility_monthly_production = data.iloc[:, [0] + list(range(45, 76))].drop_duplicates().reset_index(drop=True)

# Inventory and metering data
facility_inventory_meter = data.iloc[:, [0] + list(range(76, 94))].drop_duplicates().reset_index(drop=True)

# Add all data to a dictionary to loop over it and do high level analysis
data_dict = {
    "facilities": facilities,
    "facility_capacity": facility_capacity,
    "facility_total_production": facility_total_production,
    "facility_monthly_production": facility_monthly_production,
    # "facility_inventory_meter": facility_inventory_meter
}
```

```python
# Facility inventory and meter looks empty
# We will drop this from our data
# but this needs to be highlighted
facility_inventory_meter.info()
```

```python
for name, df in data_dict.items():
    df.columns = df.columns.str.lower().str.replace(' ', '_').str.replace('(', '_').str.replace(')', '')
```

```python
# Number of rows for each table
for name, df in data_dict.items():
    print(name.replace("_", " ").title(), "# of rows:", df.index.size)
```

```python
def dms_to_dd(dms):
    """ Parse coordinate data and convert to degrees"""
    if pd.isna(dms):
        return
    degrees, direction = dms.split(' deg ')
    degrees = float(degrees)
    if direction in ['S', 'W']:
        degrees = -degrees
    return degrees
```

```python
data_dict["facilities"]['latitude']
```

```python
# Data manipulation
data_dict["facilities"]["latitude"] = data_dict["facilities"]["latitude"].apply(dms_to_dd)
data_dict["facilities"]["longitude"] = data_dict["facilities"]["longitude"].apply(dms_to_dd)
```

```python
# Peep into each tables
for name, df in data_dict.items():
    print("Displaying", name)
    display(df.head(10))
```

```python
# This gets the id for the highest value per column
# Only returns na if there is nothing in that column
max_indices = facility_total_production.idxmax()

for column, idx in max_indices.items():
    if pd.isna(idx) or column == 'id':
        continue

    # Get the index from the facility production table
    facility_id = facility_total_production.iloc[int(idx)].id
    value = facility_total_production.iloc[int(idx)][column]

    # If the max is 0, we skip
    if value == 0.0:
        continue

    # Get the facility name
    facility = facilities.query(f"id == '{facility_id}'")[['name', 'current_operator']]

    name = facility['name'].values[0].title()
    operator = facility['current_operator'].values[0]

    print(f"Highest Facility For {column.replace('_', ' ').title()}: {value:,.2f} at {name}, operated by {operator}")
```
